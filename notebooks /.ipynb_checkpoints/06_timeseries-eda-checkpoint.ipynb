# 06 — Time-series EDA

Three checks, each one able to change a decision in `features.py`. Run before the builder.

**Deliberately not here:** ADF / KPSS / differencing / ACF–PACF / STL. That toolkit belongs to
ARIMA-family models, where the only input is the target's own history and you difference until the
series is stationary. This project fits *supervised regression on time-ordered rows* — sun geometry
and weather are explicit input columns, so the daily and seasonal non-stationarity is modelled
directly by `cos_zenith`, `hour_sin/cos` and `doy_sin/cos`. An ADF test here would print
"non-stationary" and change nothing.

| check | question | what it decides |
|---|---|---|
| 1 | does the test set look like the training set? | how results are reported |
| 2 | is lag-24 actually available? | whether the `history` block gets built at all |
| 3 | how different are the 7 clients? | how hard federated averaging will be |

```python
import json
import pandas as pd
from pathlib import Path

from solarfl.labels.capacity import add_capacity_factor

ROOT = Path().cwd().parent
splits = json.loads((ROOT / "configs/splits.json").read_text())
labels = json.loads((ROOT / "configs/station_labels.json").read_text())


def load(sid: str, capacity_kw: float) -> pd.DataFrame:
    """One client, sorted, with the target attached. Never drops rows."""
    df = pd.read_csv(ROOT / f"data/processed/client/{sid}.csv", parse_dates=["measured_ts"])
    df = add_capacity_factor(df, capacity_kw)
    return df.sort_values("measured_ts").reset_index(drop=True)


# daylight per features.yaml: is_daylight = terrestrial_radiation > 10, i.e. the sun is up
# by geometry. Filtering on measured irradiance instead would mix "night" with "very cloudy".
def daylight(df: pd.DataFrame) -> pd.DataFrame:
    return df[df["terrestrial_radiation"] > 10]
```

## Check 1 — does the test set look like the training set?

Splits are temporal (D-frozen in `configs/splits.json`), so train is the darker end of the calendar
and test is the brightest. Night rows are excluded here or the answer would just track how much of
each split is darkness.

```python
from IPython.display import display

rows = []
for sid, info in labels.items():
    df = daylight(load(sid, info["capacity_kw"]))
    s = splits["clients"][sid]

    # same boundaries as solarfl.data.splits._assign, expressed as bins over timestamps
    part = pd.cut(df["measured_ts"],
                  [pd.Timestamp.min, pd.Timestamp(s["train_end"]),
                   pd.Timestamp(s["val_end"]), pd.Timestamp.max],
                  labels=["train", "val", "test"])
    for p, g in df.groupby(part, observed=True):
        rows.append({"station": info["label"], "split": p,
                     "mean_cf": g["capacity_factor"].mean(), "n": len(g)})


split1 = pd.DataFrame(rows)

mean_cf = split1.pivot(index="station", columns="split", values="mean_cf").round(3)
counts = split1.pivot(index="station", columns="split", values="n")

print("📊 Mean Capacity Factor (Daylight Rows Only)")
display(
    mean_cf.style
        .format("{:.3f}")
        .set_properties(**{"text-align": "center"})
        .set_table_styles([
            {"selector": "th", "props": [("text-align", "center")]}
        ])
)

print("📊 Number of Daylight Rows per Split")
display(
    counts.style
        .format("{:,}")
        .set_properties(**{"text-align": "center"})
        .set_table_styles([
            {"selector": "th", "props": [("text-align", "center")]}
        ])
)
```

```python
# Ratio makes the size of the distribution shift explicit
p = split1.pivot(index="station", columns="split", values="mean_cf")

ratio = (p["test"] / p["train"]).rename("Test / Train").round(2).to_frame()

print("📊 Test-to-Train Mean Capacity Factor Ratio")
display(
    ratio.style
        .format("{:.2f}")
        .set_properties(**{"text-align": "center"})
        .set_table_styles([
            {"selector": "th", "props": [("text-align", "center")]}
        ])
)
```

### Finding 1 — every client is tested on brighter weather than it was trained on

`test` mean capacity factor is **1.45–2.0× `train`** for all 7 stations (S6 is the extreme:
0.107 → 0.215), and `val` sits close to `test`. Nothing is broken — this is the calendar, not a bug — but it fixes two reporting rules:

- **No single headline MAE.** An absolute error computed on May rows is not comparable to one
  computed on January rows. Report per split, and alongside a scale-free number
  (nMAE = MAE / mean capacity factor of that split).
- **Baselines stay valid.** Same-hour-yesterday persistence faces the identical shift, so the
  model-vs-baseline comparison is unaffected. This is exactly why the baseline is the reference
  point rather than an absolute error target.

## Check 2 — is lag-24 actually available?

`CLAUDE.md` warns of ~21% inter-day missingness and predicts lag-24 is "often unavailable". If that
holds, the `history` block in `features.yaml` is mostly NaN and not worth building.

Two things are measured separately: the row 24 hours earlier **existing**, and its
`capacity_factor` being **non-null** (it is blanked where cf > 1.0).

```python
avail = []
for sid, info in labels.items():
    df = load(sid, info["capacity_kw"])
    idx = pd.DatetimeIndex(df["measured_ts"])

    has_row = (idx - pd.Timedelta(hours=24)).isin(idx).mean()
    lagged = df.set_index("measured_ts")["capacity_factor"].reindex(idx - pd.Timedelta(hours=24))
    avail.append({"station": info["label"], "n_rows": len(df),
                  "row_at_lag24": has_row, "usable_cf_at_lag24": lagged.notna().mean()})

display(pd.DataFrame(avail).set_index("station").round(3))
```

```python
# why the 21% figure and the number above can both be true:
# coverage of the hourly grid is poor, but the holes are structured.
diag = []
for sid, info in labels.items():
    df = load(sid, info["capacity_kw"])
    ts = pd.DatetimeIndex(df["measured_ts"])
    span_h = (ts.max() - ts.min()) / pd.Timedelta("1h") + 1

    days = pd.Series(pd.to_datetime(sorted(set(ts.normalize()))))
    day_step = days.diff().dt.days.dropna()
    diag.append({"station": info["label"],
                 "hourly_grid_coverage": len(df) / span_h,
                 "days_present": len(days),
                 "hour_of_day_min": ts.hour.min(), "hour_of_day_max": ts.hour.max(),
                 "multi_day_outages": int((day_step > 1).sum()),
                 "days_lost_to_outages": int((day_step[day_step > 1] - 1).sum())})

display(pd.DataFrame(diag).set_index("station").round(3))
```

### Finding 2 — lag-24 is fine. Build the history block.

**96–98% of rows have a usable `capacity_factor` 24 hours earlier.** The `CLAUDE.md` gotcha is
measuring something real but different, and the diagnostic above shows why both are true:

1. Coverage of the full hourly grid is only ~45–84%, but the missing hours are *the same deep-night
   hours every day*. A row and its 24-hour-earlier partner sit at the same hour of day, so they are
   either both present or both absent.
2. Whole-day loss comes as **2–3 contiguous outages per station**, not scattered single days. A
   contiguous outage costs one day of lag-24 at its trailing edge; 30 scattered days would have cost 30.

So: build `history: [capacity_factor]` at lag 24 as specified, and accept a ~2–4% row loss from the
NaN drop. The `CLAUDE.md` gotcha line needs correcting — flagging, not editing.

## Check 3 — how different are the 7 clients?

The research question is whether federated learning matches centralized training. That depends
entirely on whether the clients share a physical response to sunlight. Each column below is one
station's **response curve**: mean capacity factor at a given level of measured plane-of-array
irradiance. Read across a row — same row means same sunlight, so any spread is the station, not
the weather.

```python
bins = [10, 200, 400, 600, 800, 1400]

curves, counts = {}, {}
for sid, info in labels.items():
    df = load(sid, info["capacity_kw"])
    b = pd.cut(df["global_tilted_irradiance"], bins)
    g = df.groupby(b, observed=True)["capacity_factor"]
    curves[info["label"]] = g.mean()
    counts[info["label"]] = g.size()

curves = pd.DataFrame(curves)
print("mean capacity_factor by irradiance bin")
display(curves.round(3))
print()
print("rows per bin")
display(pd.DataFrame(counts))
```

```python
# spread within a row = how non-IID the clients are at identical sunlight
spread = pd.DataFrame({"min": curves.min(axis=1), "max": curves.max(axis=1)})
spread["max / min"] = (spread["max"] / spread["min"]).round(2)
display(spread.round(3))
print()
print("rated capacity (kW) vs mean cf in the brightest bin:")
display(pd.DataFrame({"capacity_kw": {i["label"]: i["capacity_kw"] for i in labels.values()},
                    "cf_800_1400": curves.iloc[-1]}).sort_values("capacity_kw"))
```

### Finding 3 — the clients are strongly non-IID, and it is a *level* difference, not a *shape* one

At identical irradiance the stations differ by up to ~2×, and the ordering is stable across every
bin — this is not noise in a thin bin. Two things worth stating before any model runs:

- **The gap tracks rated capacity.** The 44 kW stations (S1, S2) convert a much larger fraction of
  their rating than the 440–605 kW stations (S4, S5, S6). Bigger plants lose more to string
  mismatch, inverter loading and row-to-row shading, so this is physically expected rather than a
  data fault.
- **S7 saturates.** Its curve is the steepest early and then flattens at the top
  (600–800 → 800–1400 barely moves), the signature of a plant hitting an export or inverter limit.
  It is also the only client with a full year of data.

**Prediction to hold the FL experiment to:** because the curves are near-parallel — the same shape
at different levels — the divergence is mostly a per-client *offset*. A single global FedAvg model
should overpredict the low group (S4, S6, S5) and underpredict the high group (S2, S1), yet a small
amount of per-client personalisation should recover most of the gap. If personalised FL lands close
to centralized while pure FedAvg does not, that is the predicted outcome, not a surprise.

This prediction is a judgment call and **needs a D-entry** in `docs/framing.md`.

## What this changes in `features.py`

- **Nothing structural.** Check 2 clears the `history` block: lag 24 stays, per H = 24 (D-012).
- Drop rows with NaN in any required column, as planned — expected cost ~2–4%, not the ~21% feared.
- Keep `is_daylight` as a *feature*, not a filter, so the builder still returns every row and row
  counts stay reconcilable against `splits.json`.
- Reporting, not building: metrics go out per split with a scale-free companion (Finding 1), and
  per client (Finding 3).
