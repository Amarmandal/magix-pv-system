# 07 — Results figures

Publication figures for the paper, built from the three frozen result tables
(`results/baselines_val.csv`, `results/fedavg_val.csv`, `results/fedprox_val.csv`).
Figures are saved to `results/graph/` as PDF (vector master for the paper) and
300-dpi PNG (slides / README).

**Design decisions** (each a candidate for a D-entry):

1. **Color follows the training regime, never the panel** — blue = local,
   orange = centralized, aqua = FedAvg, violet = FedProx, gray = persistence —
   so a reader can track "federated" across every figure by hue alone. The
   palette is colorblind-validated (worst adjacent-pair CVD ΔE 9.2, all
   checks pass).
2. **Persistence is a reference line, not a bar.** It is the baseline every
   model must beat (project invariant), so it spans each client group as a
   dashed tick instead of competing for area as a fifth bar.
3. **Figure 1 uses the E=1 federated arms** — one local epoch per round does
   the same gradient work as one centralized epoch, so the comparison with
   the centralized baseline is like-for-like (D-018). The E=5 arms appear
   only in Figure 2, compared against each other.
4. **Figure 3 macro-averages across stations** (unweighted mean of the 7
   per-client MAEs) rather than pooling rows, so S7's larger validation set
   (593 rows vs ~250) does not dominate the summary. The per-client dots
   show the spread the mean hides.
5. RMSE and skill are left to the paper's tables — MAE is the selection
   metric everywhere else in the project, so the figures stay on one metric.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

ROOT = Path.cwd().parent
RESULTS = ROOT / "results"
GRAPH = RESULTS / "graph"
GRAPH.mkdir(parents=True, exist_ok=True)

# Colorblind-validated categorical palette; color follows the regime in every
# figure so identity is consistent across the whole paper.
C = {
    "local": "#2a78d6",        # blue
    "central": "#eb6834",      # orange
    "fedavg": "#1baf7a",       # aqua
    "fedprox": "#4a3aa7",      # violet
    "persistence": "#898781",  # gray: reference, not a competitor
}
INK, MUTED, GRID, BASE = "#0b0b0b", "#52514e", "#e1e0d9", "#c3c2b7"

plt.rcParams.update({
    # journal-ready output: 300-dpi rasters; Type 42 (TrueType) fonts keep
    # figure text searchable/editable in the PDF instead of outlined paths
    "savefig.dpi": 300, "pdf.fonttype": 42, "ps.fonttype": 42,
    "font.size": 8, "axes.titlesize": 8.5, "axes.labelsize": 8,
    "xtick.labelsize": 7.5, "ytick.labelsize": 7.5, "legend.fontsize": 7.5,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.edgecolor": BASE, "axes.linewidth": 0.8,
    "xtick.color": MUTED, "ytick.color": MUTED,
    "axes.labelcolor": INK, "text.color": INK,
    "figure.constrained_layout.use": True,
})


def save(fig, name):
    for ext in ("pdf", "png"):
        fig.savefig(GRAPH / f"{name}.{ext}")
    print(f"saved -> results/graph/{name}.pdf + .png")


baselines = pd.read_csv(RESULTS / "baselines_val.csv")
fedavg = pd.read_csv(RESULTS / "fedavg_val.csv")
fedprox = pd.read_csv(RESULTS / "fedprox_val.csv")

CLIENTS = sorted(baselines["client"].unique())
VARIANTS = ["past", "perfect"]
VLABEL = {"past": "Past-weather variant", "perfect": "Perfect-forecast variant"}
NAMES = {"local": "Local (per-station)", "central": "Centralized (pooled)",
         "fedavg": "FedAvg", "fedprox": "FedProx",
         "persistence": "Persistence"}

# One tidy frame of the four MLP regimes; federated arms at E=1 so the
# gradient budget matches the centralized epochs (like-for-like, D-018).
cols = ["client", "variant", "regime", "mae"]
mlp = baselines.query("model == 'mlp'")[cols]
fed1 = fedavg[cols]
prox1 = fedprox.query("regime == 'fedprox' and local_epochs == 1")[cols]
regimes_e1 = pd.concat([mlp, fed1, prox1], ignore_index=True)
regimes_e1.groupby(["variant", "regime"])["mae"].mean().round(4)
```

## Figure 1 — per-client MAE, all training regimes

The headline figure: does federated training match centralized without pooling
raw data? Grouped bars per station, persistence as a dashed reference tick
spanning each group. Bars above the gray tick fail the project's minimum bar
(beat same-hour-yesterday).

```python
REGIME_ORDER = ["local", "central", "fedavg", "fedprox"]

fig, axes = plt.subplots(1, 2, figsize=(7.0, 3.0), sharey=True)
for ax, variant in zip(axes, VARIANTS):
    sub = regimes_e1[regimes_e1["variant"] == variant]
    sns.barplot(data=sub, x="client", y="mae", hue="regime",
                hue_order=REGIME_ORDER, order=CLIENTS, palette=C,
                saturation=1, width=0.8, legend=False, ax=ax)
    pers = (baselines[(baselines["model"] == "persistence")
                      & (baselines["variant"] == variant)]
            .set_index("client")["mae"])
    for i, cl in enumerate(CLIENTS):
        ax.hlines(pers[cl], i - 0.44, i + 0.44, color=C["persistence"],
                  lw=1.3, linestyle=(0, (3, 1.5)), zorder=4)
    ax.set_title(VLABEL[variant], loc="left")
    ax.set_xlabel(None)
    ax.set_ylabel("Validation MAE (capacity factor)"
                  if variant == "past" else None)
    ax.grid(axis="y", color=GRID, lw=0.6)
    ax.set_axisbelow(True)

handles = [Patch(facecolor=C[r], label=NAMES[r]) for r in REGIME_ORDER]
handles.append(Line2D([], [], color=C["persistence"], lw=1.3,
                      linestyle=(0, (3, 1.5)),
                      label="Persistence (same hour yesterday)"))
fig.legend(handles=handles, ncols=5, frameon=False,
           loc="outside upper center")
save(fig, "fig1_mae_by_client")
plt.show()
```

## Figure 2 — FedProx vs FedAvg (the non-IID question)

Dumbbell per station: the connector runs from FedAvg (aqua) to FedProx
(violet), so its direction shows per-client gain or loss from the proximal
term. Panels split by weather variant (rows) and local epochs E (columns) —
drift accumulates over local epochs, so E=5 is where FedProx has room to act.
Each panel's title reports the μ selected on pooled val MAE for that setting.

```python
fig, axes = plt.subplots(2, 2, figsize=(7.0, 4.6), sharex=True, sharey=True)
ypos = np.arange(len(CLIENTS))[::-1]  # S1 on top, reading order
for row, variant in enumerate(VARIANTS):
    for col, E in enumerate((1, 5)):
        ax = axes[row, col]
        sub = fedprox[(fedprox["variant"] == variant)
                      & (fedprox["local_epochs"] == E)]
        fa = sub[sub["regime"] == "fedavg"].set_index("client")["mae"]
        fp = sub[sub["regime"] == "fedprox"].set_index("client")["mae"]
        best_mu = sub.loc[sub["regime"] == "fedprox", "mu"].iloc[0]
        for yv, cl in zip(ypos, CLIENTS):
            ax.plot([fa[cl], fp[cl]], [yv, yv], color=BASE, lw=1.2, zorder=1)
        # white marker edges keep the pair readable where the two nearly
        # coincide (e.g. S4 in the perfect / E=1 panel)
        ax.scatter(fa[CLIENTS], ypos, s=30, color=C["fedavg"],
                   edgecolor="white", lw=0.8, zorder=2)
        ax.scatter(fp[CLIENTS], ypos, s=30, color=C["fedprox"],
                   edgecolor="white", lw=0.8, zorder=3)
        ax.set_yticks(ypos, CLIENTS)
        ax.set_title(f"{VLABEL[variant]}, E = {E}"
                     f" (best $\\mu$ = {best_mu:g})", loc="left")
        ax.grid(axis="x", color=GRID, lw=0.6)
        ax.set_axisbelow(True)
for ax in axes[1]:
    ax.set_xlabel("Validation MAE (capacity factor)")

handles = [Line2D([], [], marker="o", ls="", ms=6, mec="white",
                  color=C[r], label=NAMES[r]) for r in ("fedavg", "fedprox")]
fig.legend(handles=handles, ncols=2, frameon=False,
           loc="outside upper center")
save(fig, "fig2_fedprox_vs_fedavg")
plt.show()
```

## Figure 3 — summary: macro-averaged MAE per regime

One compact answer per regime and variant: bars are the unweighted mean of
the 7 per-station MAEs (so S7's larger validation set does not dominate);
black dots are the individual stations, showing the spread behind each mean.
The rightmost dot in every regime is S7 — the hardest station.

```python
pers_rows = (baselines.query("model == 'persistence'")
             .assign(regime="persistence")[cols])
df3 = pd.concat([pers_rows, regimes_e1], ignore_index=True)
ORDER3 = ["persistence", "local", "central", "fedavg", "fedprox"]

fig, axes = plt.subplots(1, 2, figsize=(7.0, 2.4), sharex=True, sharey=True)
y = np.arange(len(ORDER3))[::-1]  # persistence on top: reference first
ymap = dict(zip(ORDER3, y))
for ax, variant in zip(axes, VARIANTS):
    sub = df3[df3["variant"] == variant]
    mean = sub.groupby("regime")["mae"].mean().reindex(ORDER3)
    ax.barh(y, mean.values, height=0.62, color=[C[r] for r in ORDER3],
            zorder=2)
    ax.scatter(sub["mae"], sub["regime"].map(ymap), s=12, color=INK,
               edgecolor="white", lw=0.5, zorder=3)
    for yv, v in zip(y, mean.values):  # direct labels: 5 per panel, no clutter
        ax.text(0.004, yv, f"{v:.3f}", va="center", ha="left",
                color="white", fontsize=7)
    ax.set_yticks(y, [NAMES[r] for r in ORDER3])
    ax.set_title(VLABEL[variant], loc="left")
    ax.grid(axis="x", color=GRID, lw=0.6)
    ax.set_axisbelow(True)
    ax.set_xlabel("Validation MAE, macro-averaged over stations")
save(fig, "fig3_summary_macro_mae")
plt.show()
```

## What to review

- Figure 1 shows the **E=1** federated arms; the E=5 story lives only in
  Figure 2. If the paper's narrative leads with E=5, swap the source rows.
- Figure 3's macro-average weights every station equally — the alternative
  (pooling all validation rows) would weight S7 ~2.4× the others. Needs a
  D-entry either way.
- MAE only in figures; RMSE and skill are in the paper's tables.
- The aqua (FedAvg) bar is the lightest color on white; the in-figure value
  labels (Fig 3) and the paper's tables are the mitigation.
